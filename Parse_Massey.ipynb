```python
#1991 not loading properly. Saved as an html file and parsed using the html parser. 

from Import_Massey_html import HTMLParser
import re
import pandas as pd


class _MasseyTableParser(HTMLParser):
    def __init__(self) -> None:
        super().__init__()
        self.in_table = False
        self.in_row = False
        self.in_cell = False
        self.in_detail = False
        self.main_parts: list[str] = []
        self.detail_parts: list[str] = []
        self.current_row: list[tuple[str, str]] = []
        self.rows: list[list[tuple[str, str]]] = []

    def handle_starttag(self, tag: str, attrs: list[tuple[str, str | None]]) -> None:
        attr_map = {k: (v or "") for k, v in attrs}
        classes = set(attr_map.get("class", "").split())
        if tag == "table" and attr_map.get("id") == "SHCtable":
            self.in_table = True
        elif self.in_table and tag == "tr" and "bodyrow" in classes:
            self.in_row = True
            self.current_row = []
        elif self.in_row and tag == "td":
            self.in_cell = True
            self.in_detail = False
            self.main_parts = []
            self.detail_parts = []
        elif self.in_cell and tag == "div" and "detail" in classes:
            self.in_detail = True

    def handle_endtag(self, tag: str) -> None:
        if tag == "div" and self.in_detail:
            self.in_detail = False
        elif tag == "td" and self.in_cell:
            main_text = re.sub(r"\s+", " ", "".join(self.main_parts)).strip()
            detail_text = re.sub(r"\s+", " ", "".join(self.detail_parts)).strip()
            self.current_row.append((main_text, detail_text))
            self.in_cell = False
            self.in_detail = False
        elif tag == "tr" and self.in_row:
            if self.current_row:
                self.rows.append(self.current_row)
            self.in_row = False
        elif tag == "table" and self.in_table:
            self.in_table = False

    def handle_data(self, data: str) -> None:
        if not self.in_cell:
            return
        if self.in_detail:
            self.detail_parts.append(data)
        else:
            self.main_parts.append(data)


def parse_Massey_ratings(html_path: str, year: int = 1991) -> pd.DataFrame:
    with open(html_path, "r", encoding="utf-8", errors="replace") as f:
        parser = _MasseyTableParser()
        parser.feed(f.read())

    parsed_data = []
    for cells in parser.rows:
        team_name, conference = cells[0]
        row_data = [team_name, conference]
        for main_val, detail_val in cells[1:]:
            row_data.append(main_val)
            row_data.append(detail_val)
        parsed_data.append(row_data)

    columns = [
        "Team", "Conference", "Record", "Win Pct",
        "Rank 1", "Detail 1", "Rank 2", "Detail 2",
        "Rank 3", "Detail 3", "Rank 4", "Detail 4",
        "Rank 5", "Detail 5", "Rank 6", "Detail 6",
        "Rank 7", "Detail 7",
    ]
    df = pd.DataFrame(parsed_data).reindex(columns=range(len(columns)), fill_value="")
    df.columns = columns
    df["Year"] = year
    return df


# --- 1. Load existing full CSV & parse 1991 HTML ---
csv_path = "massey_ratings.csv"  # Change to "massey_ratings_2.csv" if needed
html_path = "Massey1991.html"

base_df = pd.read_csv(csv_path, dtype=str)
df_1991 = parse_Massey_ratings(html_path, year=1991).astype(str)

# Ensure exact column order and presence of any extra/renamed columns (e.g., 'Power')
for col in base_df.columns:
    if col not in df_1991.columns:
        df_1991[col] = ""
df_1991 = df_1991[base_df.columns]

# --- 2. Replace any existing 1991 rows, sort chronologically by Year and Rank 1 ---
merged_df = pd.concat(
    [base_df[base_df["Year"] != "1991"], df_1991],
    ignore_index=True,
)
merged_df["_year_num"] = pd.to_numeric(merged_df["Year"], errors="coerce")
merged_df["_rank1_num"] = pd.to_numeric(merged_df["Rank 1"], errors="coerce")
merged_df = (
    merged_df.sort_values(["_year_num", "_rank1_num"], na_position="last")
    .drop(columns=["_year_num", "_rank1_num"])
    .reset_index(drop=True)
)

# --- 3. Save updated CSV (and optional .xlsx if you want a spreadsheet workbook) ---
merged_df.to_csv(csv_path, index=False)
# merged_df.to_excel("massey_ratings_full.xlsx", index=False)

print(f"Total rows: {len(merged_df)} | 1991 rows: {(merged_df['Year'] == '1991').sum()}")
```
