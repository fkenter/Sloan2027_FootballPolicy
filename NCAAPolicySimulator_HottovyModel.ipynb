```python
import random
import math
import numpy as np
import pandas as pd
from collections import defaultdict
from scipy.stats import norm
import matplotlib.pyplot as plt
import os

# ==========================================
# 1. CONFIGURATION & CONSTANTS
# ==========================================

NUM_TEAMS = 134
SEASONS = 50
MIN_POWER = 25
MAX_POWER = 85
LEAGUE_AVG_POWER = 55
ST_DEV_NOISE = 4.70735
TRANSITION_ALPHA = 0.9
HISTORICAL_REVERSION = 0.1
LEAGUE_WEIGHT = 1.0 - TRANSITION_ALPHA - HISTORICAL_REVERSION

LEAGUE_TIER_BONUS = 15
STD_GAME = 14
PTS_WIN = 1

USE_CONF_COEFFS = False  # Set to True to use Conference Coeffs, False for W/O Conferences

# ==========================================
# 2. DATA SETUP (EXACT ORIGINAL TEAMS)
# ==========================================

def get_modern_conferences_134():
    """
    Returns the 2025 Conference Alignment using EXACT ORIGINAL TEAM IDs
    as provided in the NCAA Promotion Simulator.
    """
    conf = {}
    
    conf["Southeastern"] = [
        "ALA", "ARK", "AUB", "FLA", "Georgia", "Kentucky", "LSU", 
        "Mississippi", "Mississippi State", "Missouri", "Oklahoma", 
        "South Carolina", "Tennessee", "Texas", "Texas A&M", "Vanderbilt"
    ]
    conf["Big_10"] = [
        "Illinois", "Indiana", "Iowa", "Maryland", "Michigan", "Michigan State", 
        "Minnesota", "Nebraska", "Northwestern", "Ohio State", "Oregon", 
        "Penn State", "Purdue", "Rutgers", "UCLA", "USC", "Washington", "Wisconsin"
    ]
    conf["Big_12"] = [
        "ARIZ", "Arizona St.", "BAY", "BYU", "CIN", "Colorado", "Houston", 
        "Iowa State", "KANSAS", "Kansas State", "Oklahoma State", "TCU", 
        "Texas Tech", "UCF", "Utah", "West Virginia"
    ]
    conf["Atlantic_Coast"] = [
        "BC", "Cal", "CLEM", "Duke", "Florida State", "Georgia Tech", 
        "Louisville", "Miami", "N.C. State", "North Carolina", "Pittsburgh", 
        "SMU", "Stanford", "Syracuse", "Virginia", "Virginia Tech", "Wake Forest"
    ]
    conf["Pac_12"] = ["Oregon State", "Washington State"]
    conf["Mountain_West"] = [
        "AF", "Boise", "Colorado St.", "Fresno State", "Hawaii", "Nevada", 
        "NM", "San Diego State", "San Jose State", "UNLV", "Utah State", "Wyoming"
    ]
    conf["American_Athletic"] = [
        "Army", "Charlotte", "ECU", "Florida Atlantic", "Memphis", "Navy", 
        "North Texas", "Rice", "Temple", "TULANE", "Tulsa", "UAB", "USF", "UTSA"
    ]
    conf["Sun_Belt"] = [
        "APP ST", "ARK ST", "C CAR", "Georgia Southern", "Georgia St.", 
        "J. Madison", "MARSHALL", "Old Dominion", "South Alabama", 
        "Southern Miss", "Texas State", "TROY", "UL-Lafayette", "UL-Monroe"
    ]
    conf["Mid_American"] = [
        "Akron", "BALL", "BGSU", "BUF", "C. Mich.", "E. Michigan", 
        "Kent State", "Miami (Ohio)", "Northern Illinois", "Ohio", 
        "Toledo", "Western Michigan"
    ]
    conf["Conference_USA"] = [
        "FIU", "Jacksonville St.", "Kennesaw St.", "Liberty", "Louisiana Tech", 
        "Middle Tennessee", "N. Mexico St.", "Sam Houston St.", "UTEP", "Western Kentucky"
    ]
    conf["FBS_Indep"] = ["Connecticut", "UMASS", "Notre Dame"]

    all_teams = []
    for v in conf.values(): 
        all_teams.extend(v)
    return conf, all_teams

# ==========================================
# 3. CORE CLASSES (TEAM & LEAGUE)
# ==========================================

class Team:
    def __init__(self, t_id, conf_tag):
        self.id = t_id
        self.conf_tag = conf_tag
        self.power = random.uniform(MIN_POWER, MAX_POWER) 
        self.historical_power = LEAGUE_AVG_POWER
        
        # History & Tracking
        self.history_power = []
        self.history_points = []
        self.ranking_metric = 0 
        self.tier = 14 # Default low tier
        self.season_points = 0
        self.wins = 0
        self.losses = 0
        self.broadcast_count = 0

    def reset_season_stats(self):
        # Save points to history
        self.history_points.append(self.season_points)
        
        self.season_points = 0
        self.wins = 0
        self.losses = 0
        self.broadcast_count = 0

class League:
    def __init__(self):
        self.conf_structure, self.team_ids = get_modern_conferences_134()
        self.teams = [Team(tid, c) for c, tlist in self.conf_structure.items() for tid in tlist]
        self.teams_dict = {t.id: t for t in self.teams}
        self.set_permanent_historical_power()

    def get_team(self, t_id):
        return self.teams_dict.get(t_id)

    def set_permanent_historical_power(self):
        """Sets a permanent historical baseline for each team based on initial conference strength."""
        conf_avgs = {}
        for conf, members in self.conf_structure.items():
            powers = [self.get_team(t).power for t in members]
            conf_avgs[conf] = sum(powers) / len(powers) if powers else MIN_POWER
        
        for t in self.teams:
            t.historical_power = conf_avgs[t.conf_tag]

    

    def update_all_powers(self):
        # Smoothed Regime-Switching AR(3) Parameters
        PHI_1 = 0.55
        PHI_2 = 0.20
        PHI_3 = 0.10
        ST_DEV_NOISE = 4.5
        REGIME_VOLATILITY = 5.0
        ALPHA_SMOOTHING = 0.4
        LEAGUE_AVG_POWER = 55
        
        """Updates true power using the Smoothed Regime-Switching AR(3) model."""
        current_powers = []
        
        for t in self.teams:
            # 1. Initialize State Variables if missing
            if not hasattr(t, 'z_history'):
                t.z_history = [0.0, 0.0, 0.0] # The AR(3) memory
            if not hasattr(t, 'regime_target'):
                t.regime_target = np.random.normal(0, REGIME_VOLATILITY)
                t.regime_years_left = np.random.randint(5, 11)
            if not hasattr(t, 'smoothed_baseline'):
                t.smoothed_baseline = t.regime_target
                
            # 2. Check for a Regime Shift (Coaching Change / Structural Break)
            t.regime_years_left -= 1
            if t.regime_years_left <= 0:
                # Draw a new 5-10 year era baseline
                t.regime_target = np.random.normal(0, REGIME_VOLATILITY)
                t.regime_years_left = np.random.randint(5, 11)
                
            # 3. Apply Exponential Smoothing to the Baseline (multi-year transition)
            t.smoothed_baseline = (ALPHA_SMOOTHING * t.regime_target) + ((1 - ALPHA_SMOOTHING) * t.smoothed_baseline)
            
            # 4. Generate the AR(3) Fluctuations (Z_t)
            noise = np.random.normal(0, ST_DEV_NOISE)
            z_t = (PHI_1 * t.z_history[-1]) + (PHI_2 * t.z_history[-2]) + (PHI_3 * t.z_history[-3]) + noise
            
            # 5. Combine into Final Power Rating
            new_power = t.smoothed_baseline + z_t
            t.power = new_power
            current_powers.append(new_power)
            
            # Store AR state memory
            t.z_history.append(z_t)
            t.z_history.pop(0) # Keep only last 3 years
            
        # Cross-sectional Mean-Centering
        current_avg = sum(current_powers) / len(current_powers)
        correction = LEAGUE_AVG_POWER - current_avg
        current_max = max(current_powers) - correction
        current_min = min(current_powers) - correction
        
        for t in self.teams:
            t.power += correction
            t.power = max(min(t.power,MAX_POWER),MIN_POWER)

    def simulate_game(self, id1, id2):
        t1, t2 = self.get_team(id1), self.get_team(id2)
        prob = norm.cdf((t1.power - t2.power) / STD_GAME)
        if random.random() < prob:
            t1.wins += 1; t1.season_points += PTS_WIN; t2.losses += 1
            return id1
        else:
            t2.wins += 1; t2.season_points += PTS_WIN; t1.losses += 1
            return id2

# ==========================================
# 4. SCHEDULING POLICIES (WITH METRICS)
# ==========================================

class SchedulingPolicy:
    def __init__(self, league):
        self.league = league
        self.teams = league.teams
    
    def generate_schedule(self):
        raise NotImplementedError("Must be implemented by subclasses.")
        
    def assign_tiers(self):
        """Default tier assignment (None). Overridden by Tier policies."""
        pass
        
    def calculate_promotion_metrics(self):
        """
        Default promotion metric (Used by Status Quo and Status Quo + OOC).
        Metric = Conference Prestige Score (historical power) + Wins.
        """
        for t in self.teams:
            conf_score = t.historical_power # Acts as a blind "Strength of Conference" buffer
            current_score = t.season_points + conf_score
            
            # Historical Moving Average (Fixing 0-index)
            hist = t.history_points[-3:] 
            padded = [0] * (3 - len(hist)) + hist
            
            weighted_metric = (
                (padded[0] * 0.1) + 
                (padded[1] * 0.2) + 
                (padded[2] * 0.3) + 
                (current_score * 0.4)
            )
            
            t.ranking_metric = weighted_metric
            t.season_points = current_score # Overwrite for next year's history points
        
    def fill_to_12_games(self, schedule):
        """Randomly fill remaining games up to 12 per team, checking limits for BOTH teams."""
        for t in self.teams:
            curr = sum(1 for g in schedule if t.id in g)
            if curr >= 12:
                continue
                
            valid_opps = [opp for opp in self.teams if opp.id != t.id and opp.conf_tag != t.conf_tag]
            random.shuffle(valid_opps)
            
            for opp in valid_opps:
                if curr >= 12: break
                
                # Check opponent's current game count
                opp_curr = sum(1 for g in schedule if opp.id in g)
                if opp_curr >= 12: continue
                
                match = tuple(sorted((t.id, opp.id)))
                if match not in schedule:
                    schedule.append(match)
                    curr += 1
                    
        return schedule

class Policy1_StatusQuo(SchedulingPolicy):
    def generate_schedule(self):
        schedule = []
        conf_groups = defaultdict(list)
        for t in self.teams: conf_groups[t.conf_tag].append(t)
            
        for conf, members in conf_groups.items():
            n_games = 8 if len(members) >= 9 else len(members) - 1
            for t in members:
                curr_conf_games = sum(1 for g in schedule if t.id in g and any(x.id in g for x in members if x != t))
                if curr_conf_games >= n_games: continue
                
                opps = [x for x in members if x != t]
                random.shuffle(opps)
                
                for opp in opps:
                    if curr_conf_games >= n_games: break
                    
                    # Check opponent's conference game count
                    opp_conf_games = sum(1 for g in schedule if opp.id in g and any(x.id in g for x in members if x != opp))
                    if opp_conf_games >= n_games: continue
                    
                    match = tuple(sorted((t.id, opp.id)))
                    if match not in schedule:
                        schedule.append(match)
                        curr_conf_games += 1
                        
        return self.fill_to_12_games(schedule)

class Policy2_StatusQuo_PerformanceOOC(SchedulingPolicy):
    def generate_schedule(self):
        schedule = []
        sorted_teams = sorted(self.teams, key=lambda x: x.ranking_metric, reverse=True)
        
        # 1. Schedule 2 OOC games based on similar performance
        for t in sorted_teams:
            ooc_games = sum(1 for g in schedule if t.id in g and self.league.get_team(g[0]).conf_tag != self.league.get_team(g[1]).conf_tag)
            if ooc_games >= 2: continue
            
            pool = [opp for opp in sorted_teams if opp.conf_tag != t.conf_tag and opp.id != t.id]
            for opp in pool:
                if ooc_games >= 2: break
                
                # Check opponent's OOC game count
                opp_ooc_games = sum(1 for g in schedule if opp.id in g and self.league.get_team(g[0]).conf_tag != self.league.get_team(g[1]).conf_tag)
                if opp_ooc_games >= 2: continue
                
                match = tuple(sorted((t.id, opp.id)))
                if match not in schedule:
                    schedule.append(match)
                    ooc_games += 1

        # 2. Standard Conference generation
        conf_groups = defaultdict(list)
        for t in self.teams: conf_groups[t.conf_tag].append(t)
        
        for conf, members in conf_groups.items():
            n_games = 8 if len(members) >= 9 else len(members) - 1
            for t in members:
                curr_conf_games = sum(1 for g in schedule if t.id in g and self.league.get_team(g[0]).conf_tag == self.league.get_team(g[1]).conf_tag)
                if curr_conf_games >= n_games: continue
                
                opps = [x for x in members if x != t]
                random.shuffle(opps)
                
                for opp in opps:
                    if curr_conf_games >= n_games: break
                    
                    # Check opponent's conference game count
                    opp_conf_games = sum(1 for g in schedule if opp.id in g and self.league.get_team(g[0]).conf_tag == self.league.get_team(g[1]).conf_tag)
                    if opp_conf_games >= n_games: continue
                    
                    match = tuple(sorted((t.id, opp.id)))
                    if match not in schedule:
                        schedule.append(match)
                        curr_conf_games += 1
                        
        return self.fill_to_12_games(schedule)

class Policy3_ConferenceTiers(SchedulingPolicy):
    def assign_tiers(self):
        """Assigns Upper (Tier 1) and Lower (Tier 2) within each conference."""
        conf_groups = defaultdict(list)
        for t in self.teams: conf_groups[t.conf_tag].append(t)
            
        for conf, members in conf_groups.items():
            if len(members) < 4: 
                for t in members: t.tier = 1
                continue
                
            members.sort(key=lambda x: x.ranking_metric, reverse=True)
            mid = len(members) // 2
            for i, t in enumerate(members):
                t.tier = 1 if i < mid else 2

    def calculate_promotion_metrics(self):
        """Metric = Tier Bonus + Standing Points + Points Per Win."""
        
        # 1. Non-linear tier bonuses extrapolated for 2 tiers
        tier_bonuses = {
            1: 20, 2: 1
        }
        
        # 2. Standing points (index 0 is 1st place, index 8 is 9th place)
        standing_points = [30, 25, 20, 15, 12, 10, 8, 5, 2]
        
        # FIX: Group teams by BOTH conference and tier so we can determine their actual standings
        conf_tiers = defaultdict(list)
        for t in self.teams:
            conf_tiers[(t.conf_tag, t.tier)].append(t)
            
        for (conf, tier_level), teams_in_tier in conf_tiers.items():
            # Sort teams within their specific conference tier by wins to determine standing
            sorted_by_standings = sorted(teams_in_tier, key=lambda x: x.wins, reverse=True)
            
            for rank, t in enumerate(sorted_by_standings):
                # Apply tier bonus (default to 1 if somehow out of bounds)
                bonus = tier_bonuses.get(t.tier, 1)
                
                # Apply standing points (10th place and below get 0)
                standings_pts = standing_points[rank] if rank < len(standing_points) else 0
                
                # Apply 1 point per win
                win_pts = t.wins * 1
                
                # Calculate new current score
                current_score = bonus + standings_pts + win_pts
                
                # Historical Moving Average weighting
                hist = t.history_points[-3:] 
                padded = [0] * (3 - len(hist)) + hist
                weighted_metric = ((padded[0] * 0.1) + (padded[1] * 0.2) + (padded[2] * 0.3) + (current_score * 0.4))
                
                t.ranking_metric = weighted_metric
                
                # Overwrite season_points for the next year's history
                t.season_points = current_score

    def generate_schedule(self):
        schedule = []
        conf_groups = defaultdict(list)
        for t in self.teams: conf_groups[t.conf_tag].append(t)
            
        for conf, members in conf_groups.items():
            if len(members) < 4:
                # Play round robin for tiny confs
                for i in range(len(members)):
                    for j in range(i+1, len(members)):
                        schedule.append(tuple(sorted((members[i].id, members[j].id))))
                continue
                
            upper = [t for t in members if t.tier == 1]
            lower = [t for t in members if t.tier == 2]
            
            for tier in [upper, lower]:
                for i in range(len(tier)):
                    for j in range(i+1, len(tier)):
                        schedule.append(tuple(sorted((tier[i].id, tier[j].id))))
                        
                other_tier = lower if tier == upper else upper
                for t in tier:
                    random.shuffle(other_tier)
                    for opp in other_tier[:2]:
                        match = tuple(sorted((t.id, opp.id)))
                        if match not in schedule:
                            schedule.append(match)
                            
        return self.fill_to_12_games(schedule)

class Policy4_PromotionTiers(SchedulingPolicy):
    def assign_tiers(self):
        """Assigns Promotion Tiers (1 through 14) based on ranking metric."""
        sorted_teams = sorted(self.teams, key=lambda x: x.ranking_metric, reverse=True)
        for i, t in enumerate(sorted_teams):
            t.tier = (i // 10) + 1

    def calculate_promotion_metrics(self):
        """Metric = Promotion Tier Bonus + Standing Points + Points Per Win."""
        
        # 1. Non-linear tier bonuses extrapolated for 14 tiers
        tier_bonuses = {
            1: 140, 2: 130, 3: 120, 4: 115, 5: 110, 
            6: 105, 7: 100, 8: 95,  9: 90,  10: 85, 
            11: 80, 12: 75, 13: 70, 14: 65
        }
        
        # 2. Standing points (index 0 is 1st place, index 8 is 9th place)
        standing_points = [30, 25, 20, 15, 12, 10, 8, 5, 2]
        
        # Group teams by tier so we can determine their standings
        tiers = defaultdict(list)
        for t in self.teams:
            tiers[t.tier].append(t)
            
        for tier_level, teams_in_tier in tiers.items():
            # Sort teams within the tier by wins to determine their standing
            sorted_by_standings = sorted(teams_in_tier, key=lambda x: x.wins, reverse=True)
            
            for rank, t in enumerate(sorted_by_standings):
                # Apply tier bonus (default to 65 if somehow out of bounds)
                bonus = tier_bonuses.get(t.tier, 65)
                
                # Apply standing points (10th place and below get 0)
                standings_pts = standing_points[rank] if rank < len(standing_points) else 0
                
                # Apply 3 points per win
                win_pts = t.wins * 3
                
                # Calculate new current score
                current_score = bonus + standings_pts + win_pts
                
                # Historical Moving Average weighting
                hist = t.history_points[-3:] 
                padded = [0] * (3 - len(hist)) + hist
                weighted_metric = ((padded[0] * 0) + (padded[1] * 0.1) + (padded[2] * 0.4) + (current_score * 0.5))
                
                t.ranking_metric = weighted_metric
                
                # Overwrite season_points for the next year's history
                t.season_points = current_score 

    def generate_schedule(self):
        schedule = []
        divisions = defaultdict(list)
        for t in self.teams:
            divisions[t.tier].append(t)
            
        for tier_num, div_teams in divisions.items():
            for i in range(len(div_teams)):
                for j in range(i+1, len(div_teams)):
                    schedule.append(tuple(sorted((div_teams[i].id, div_teams[j].id))))
                    
        return self.fill_to_12_games(schedule)

# ==========================================
# 5. GREEDY BROADCASTER
# ==========================================

class GreedyBroadcaster:
    def __init__(self, coeffs, limits, league):
        self.coeffs = coeffs
        self.limits = limits
        self.league = league
        self.final_season_broadcasts = []

    def _predict_viewership(self, t1, t2, w, d, t, b):
        
        if not self.coeffs: return math.exp(random.uniform(2, 5)) 
        
        p1 = self.league.get_team(t1).power
        p2 = self.league.get_team(t2).power
        
        currpower = p1 + p2
        powerdiff = abs(p1 - p2)
        
        nbroadcast = self.limits['maxbroadcast'][(w,d,t)]
        ncable = self.limits['maxcable'][(w,d,t)]
        totalpower_cap = self.limits['maxpower'].get((w,d,t), 100)
        
        # Abort if slot limits too low per original Pyomo constraint
        if totalpower_cap < 50: 
            return 0
            
        fracpower = min(100 * currpower / (totalpower_cap + 0.01), 100)
        fracpower_int_ngames = fracpower * (nbroadcast + ncable)
        fracpower_int_currpower = fracpower * currpower
        
        # Base Linear terms
        c_default = self.coeffs.get('cDefault', 0)
        c_year = self.coeffs.get('cYear', 0)
        c_wk = self.coeffs.get('cWeeks', {}).get(w, 0)
        c_day = self.coeffs.get('cDays', {}).get(d, 0)
        c_tm = self.coeffs.get('cTimes', {}).get(t, 0)
        c_b  = self.coeffs.get('cBroadcasts', {}).get(b, 0)
        
        # Conference level terms
        c_team1 = self.coeffs.get('cTeams', {}).get(self.league.get_team(t1).conf_tag, 0)
        c_team2 = self.coeffs.get('cTeams', {}).get(self.league.get_team(t2).conf_tag, 0)
        
        # Miscellaneous terms
        c_misc = self.coeffs.get('cMisc', {})
        
        # Summation of the full linear model
        exponent = (
            c_default + c_year + c_wk + c_day + c_tm + c_b + c_team1 + c_team2 +
            (c_misc.get("NumGames", 0) * (nbroadcast + ncable)) +
            (c_misc.get("currpwr", 0) * currpower) +
            (c_misc.get("FractionPwr", 0) * fracpower) +
            (c_misc.get("pwrdiff", 0) * powerdiff) +
            (c_misc.get("FracPwr_int_NumGames", 0) * fracpower_int_ngames) +
            (c_misc.get("FracPwr_int_currpwr", 0) * fracpower_int_currpower) +
            c_misc.get("rivalry", 0) # Assuming baseline addition for rivalry if triggered
        )
        
        # Exponentiate to get actual viewership
        return math.exp(exponent)

    def assign_schedule(self, schedule):
        """Greedily assigns games to maximize actual viewership without forcing coverage."""
        if not self.limits: return 0 
        
        weeks = list(range(0, 15))
        days = [1, 2, 3, 4, 5, 6, 7]
        times = ['morning', 'afternoon', 'evening', 'late']
        broadcasts = ['broadcast', 'cable']
        
        valid_slots = [(w, d, t) for w in weeks for d in days for t in times 
                       if self.limits['maxbroadcast'][(w,d,t)] > 0 or self.limits['maxcable'][(w,d,t)] > 0]
        
        # 1. Evaluate all combinations
        options = []
        for g_idx, (t1, t2) in enumerate(schedule):
            for w, d, t in valid_slots:
                for b in broadcasts:
                    if self.limits[f'max{b}'][(w,d,t)] > 0:
                        score = self._predict_viewership(t1, t2, w, d, t, b)
                        # Store the combined power of the two teams so we can track it later
                        game_power = self.league.get_team(t1).power + self.league.get_team(t2).power
                        options.append({'game_idx': g_idx, 't1': t1, 't2': t2, 'w': w, 'd': d, 't': t, 'b': b, 'score': score, 'game_power': game_power})
                        
        # 2. Sort purely by highest projected viewership
        options.sort(key=lambda x: x['score'], reverse=True)
        
        assigned_games = set()
        team_weeks_played = defaultdict(set)
        
        # Track available game slots AND available power per slot
        slot_caps = {f"{b}_{w}_{d}_{t}": self.limits[f'max{b}'][(w,d,t)] for w,d,t in valid_slots for b in broadcasts}
        power_caps = {(w, d, t): self.limits['maxpower'].get((w,d,t), 100) for w,d,t in valid_slots}
        
        final_broadcasts = []
        
        # 3. Single Maximization Pass
        for opt in options:
            if opt['game_idx'] in assigned_games: continue
            
            # A team can only play one game per week
            if opt['w'] in team_weeks_played[opt['t1']] or opt['w'] in team_weeks_played[opt['t2']]: continue
            
            w, d, t, b = opt['w'], opt['d'], opt['t'], opt['b']
            slot_key = f"{b}_{w}_{d}_{t}"
            power_key = (w, d, t)
            
            # Check if there is a broadcast/cable slot available AND enough power capacity left in the slot
            if slot_caps[slot_key] > 0 and (power_caps[power_key] - opt['game_power']) >= 0:
                
                # Deduct limits
                slot_caps[slot_key] -= 1
                power_caps[power_key] -= opt['game_power']
                
                assigned_games.add(opt['game_idx'])
                team_weeks_played[opt['t1']].add(w)
                team_weeks_played[opt['t2']].add(w)
                
                self.league.get_team(opt['t1']).broadcast_count += 1
                self.league.get_team(opt['t2']).broadcast_count += 1
                
                final_broadcasts.append(opt)
                
        # Store for printing in the final season
        self.final_season_broadcasts = final_broadcasts

        if not final_broadcasts: return 0
        
        # Compute and return TOTAL actual viewership for the season
        return sum(x['score'] for x in final_broadcasts)

# ==========================================
# 6. DATA LOADERS & PRE-GENERATORS
# ==========================================

def load_initial_power_rankings(league, year_to_use=2018):
    """
    Loads power rankings from 'PwrRankings for IP.csv' and assigns them 
    to the corresponding team objects in the league.
    """
    try:
        df_pwr = pd.read_csv('PwrRankings for IP.csv')
        df_sub = df_pwr[df_pwr['Year'] == year_to_use]
        pwr_dict = pd.Series(df_sub.Power.values, index=df_sub.Team).to_dict()
        
        for team in league.teams:
            if team.id in pwr_dict:
                team.power = pwr_dict[team.id]
            else:
                team.power = MIN_POWER
    except FileNotFoundError:
        pass 
    except Exception:
        pass

def load_data_from_files():
    """Loads model coefficients from excel and slot summaries from CSV."""
    coeffs = defaultdict(dict)
    limits = {
        'maxbroadcast': defaultdict(int), 
        'maxcable': defaultdict(int),
        'maxpower': defaultdict(float)
    }
    
    try:
        # Load from the provided excel file
        df1 = pd.read_excel('with and without conferences.xlsx', sheet_name='Sheet2')
        val_col = 'Unnamed: 3' if USE_CONF_COEFFS else 'Unnamed: 4'
        
        # Create a dictionary mapping from the "Variable" column to the correct value column
        df1 = df1.dropna(subset=['Unnamed: 2'])
        var_dict = dict(zip(df1['Unnamed: 2'], df1[val_col]))
        
        def parse_float(val):
            try:
                return float(val) if not pd.isna(val) else 0.0
            except:
                return 0.0

        # Map Days (Sunday=7, Monday=1, ..., Saturday is implicitly 0 baseline)
        cDays = defaultdict(float)
        cDays[7] = parse_float(var_dict.get('Sunday'))
        cDays[1] = parse_float(var_dict.get('Monday'))
        cDays[2] = parse_float(var_dict.get('Tuesday'))
        cDays[3] = parse_float(var_dict.get('Wednesday'))
        cDays[4] = parse_float(var_dict.get('Thursday'))
        cDays[5] = parse_float(var_dict.get('Friday'))
        coeffs['cDays'] = cDays

        # Map Times
        cTimes = defaultdict(float)
        cTimes['afternoon'] = parse_float(var_dict.get('Afternoon'))
        cTimes['evening'] = parse_float(var_dict.get('Evening'))
        cTimes['late'] = parse_float(var_dict.get('Late'))
        coeffs['cTimes'] = cTimes
        
        # Map Weeks (Fall back to Mid Season for unlisted middle weeks)
        mid = parse_float(var_dict.get('Mid Season'))
        cWeeks = defaultdict(lambda: mid)
        cWeeks[0] = parse_float(var_dict.get('Week 0'))
        cWeeks[2] = parse_float(var_dict.get('Week 2'))
        cWeeks[3] = parse_float(var_dict.get('Week 3'))
        cWeeks[4] = parse_float(var_dict.get('Week 4'))
        cWeeks[5] = parse_float(var_dict.get('Week 5'))
        cWeeks[11] = parse_float(var_dict.get('Rivals Week minus 4'))
        cWeeks[12] = parse_float(var_dict.get('Rivals Week minus 3'))
        cWeeks[13] = parse_float(var_dict.get('Rivals Week minus 2'))
        cWeeks[14] = parse_float(var_dict.get('Rivals Week minus 1'))
        cWeeks[15] = parse_float(var_dict.get('Rivals Week'))
        coeffs['cWeeks'] = cWeeks
        
        # Map Broadcast
        cBroadcasts = defaultdict(float)
        cBroadcasts["broadcast"] = parse_float(var_dict.get('Over-the-Air'))
        coeffs['cBroadcasts'] = cBroadcasts
        
        # Standard Intercept / Year
        coeffs['cDefault'] = parse_float(var_dict.get('Intercept'))
        coeffs['cYear'] = parse_float(var_dict.get('Year(2019)'))
        
        # Map Conferences (These will parse as 0.0 if USE_CONF_COEFFS is False)
        cTeams = defaultdict(float)
        cTeams['American_Athletic'] = parse_float(var_dict.get('American Athletic'))
        cTeams['Atlantic_Coast'] = parse_float(var_dict.get('Atlantic Coast'))
        cTeams['Big_10'] = parse_float(var_dict.get('Big 10'))
        cTeams['Big_12'] = parse_float(var_dict.get('Big 12'))
        cTeams['Conference_USA'] = parse_float(var_dict.get('Conference USA'))
        cTeams['FBS_Indep'] = parse_float(var_dict.get('FBS Indep.'))
        cTeams['Mid_American'] = parse_float(var_dict.get('Mid American'))
        cTeams['Mountain_West'] = parse_float(var_dict.get('Mountain West'))
        cTeams['Pac_12'] = parse_float(var_dict.get('Pac 12'))
        cTeams['Southeastern'] = parse_float(var_dict.get('Southeastern'))
        cTeams['Sun_Belt'] = parse_float(var_dict.get('Sun Belt'))
        coeffs['cTeams'] = cTeams
        
        # Map Misc Interactions
        cMisc = defaultdict(float)
        cMisc['NumGames'] = parse_float(var_dict.get('Games On'))
        cMisc['currpwr'] = parse_float(var_dict.get('Game Power'))
        cMisc['FractionPwr'] = parse_float(var_dict.get('Fraction Power'))
        cMisc['pwrdiff'] = parse_float(var_dict.get('Power Diff'))
        cMisc['FracPwr_int_NumGames'] = parse_float(var_dict.get('Fraction Power x Games On'))
        cMisc['FracPwr_int_currpwr'] = parse_float(var_dict.get('Fraction Power x Game Power'))
        cMisc['rivalry'] = parse_float(var_dict.get('Rivalry'))
        coeffs['cMisc'] = cMisc

    except Exception as e:
        print(f"Warning: 'with and without conferences.xlsx' failed to load ({e}). Using fallback math.")
        pass 
        
    try:
        df2 = pd.read_csv('wk_day_slot_summary_NO_FCS.csv')
        for i in range(len(df2)):
            w, d, slot = df2["Week"][i], df2["NumDay"][i], df2["Slot"][i]
            limits['maxbroadcast'][(w,d,slot)] = df2["Num Broadcast"][i]
            limits['maxcable'][(w,d,slot)] = df2["Num Cable"][i]
            limits['maxpower'][(w,d,slot)] = df2["Total Power"][i]
    except FileNotFoundError:
        print("Warning: 'wk_day_slot_summary_NO_FCS.csv' not found. Creating dummy limits.")
        for w in range(15):
            for d in [6, 7]:
                for t in ['afternoon', 'evening']:
                    limits['maxbroadcast'][(w,d,t)] = 2
                    limits['maxcable'][(w,d,t)] = 4
                    limits['maxpower'][(w,d,t)] = 100
                    
    return coeffs, limits

def pregenerate_power_trajectories():
    """
    Calculates the exact True Power for every team across all seasons independently.
    This guarantees every policy operates on the exact same underlying team strength.
    """
    base_league = League()
    load_initial_power_rankings(base_league, year_to_use=2018)
    base_league.set_permanent_historical_power()
    
    historical_baselines = {t.id: t.historical_power for t in base_league.teams}
    power_trajectories = {}
    
    for season in range(1, SEASONS + 1):
        # Save power for the current season
        power_trajectories[season] = {t.id: t.power for t in base_league.teams}
        # Advance power mathematically for the next season
        base_league.update_all_powers()
        
    return power_trajectories, historical_baselines

# ==========================================
# 7. MAIN EXECUTION
# ==========================================

def main():
    coeffs, limits = load_data_from_files()
    
    # 1. Generate the fixed 50-year power timeline
    print("Pre-generating fixed 50-year power trajectories to ensure simulation parity...")
    power_trajectories, historical_baselines = pregenerate_power_trajectories()
    
    policies = {
        "Status Quo": Policy1_StatusQuo,
        "Status Quo + OOC Performance": Policy2_StatusQuo_PerformanceOOC,
        "Conference Tiers": Policy3_ConferenceTiers,
        "Promotion Tiers": Policy4_PromotionTiers
    }
    
    results = {}
    viewership_distributions = {}  # For the Box & Whisker plot
    
    # Tracking for the line graph
    p1_top4_history = [] 
    p4_top4_history = [] 
    
    for policy_name, PolicyClass in policies.items():
        print(f"\n========================================================")
        print(f"--- Running Simulation for: {policy_name} ---")
        league = League()
        broadcaster = GreedyBroadcaster(coeffs, limits, league)
        total_viewership_tracker = []
        
        # Synchronize Historical Baselines so Tier Bonus calculations align
        for t in league.teams:
            t.historical_power = historical_baselines[t.id]
        
        for season in range(1, SEASONS + 1):
            
            # 1. Apply the exact pre-generated power for this season
            for t in league.teams:
                t.power = power_trajectories[season][t.id]
                t.history_power.append(t.power) # Maintain history for metric calcs
            
            # 2. Instantiate Policy and Assign Tiers
            policy = PolicyClass(league)
            policy.assign_tiers()
            
            # --- GRAPH 1 TRACKING LOGIC ---
            # Track Status Quo top 4 conferences
            if policy_name == "Status Quo":
                conf_powers = defaultdict(list)
                for t in league.teams:
                    conf_powers[t.conf_tag].append(t.power)
                avg_conf_powers = [np.mean(powers) for powers in conf_powers.values()]
                avg_conf_powers.sort(reverse=True)
                # Average of the top 4 conferences
                p1_top4_history.append(np.mean(avg_conf_powers[:4]))
                
            # Track Promotion Tiers top 4 tiers
            elif policy_name == "Promotion Tiers":
                tier_powers = defaultdict(list)
                for t in league.teams:
                    tier_powers[t.tier].append(t.power)
                # Tiers 1 through 4 are the top 4
                tier_avgs = [np.mean(tier_powers[tier]) for tier in [1, 2, 3, 4] if tier in tier_powers]
                p4_top4_history.append(np.mean(tier_avgs))
            # ------------------------------

            # 3. Print Team Powers before simulating the FINAL season
            if season == SEASONS:
                print(f"\n--- Team Powers Going Into Final Season ({policy_name}) ---")
                sorted_teams = sorted(league.teams, key=lambda x: x.power, reverse=True)
                for t in sorted_teams:
                    print(f"{t.id:<18} ({t.conf_tag:<18}) | Power: {t.power:>5.2f} | Tier: {t.tier:>2} | Metric: {t.ranking_metric:>6.2f}")

            # 4. Generate Schedule & Simulate Games
            schedule = policy.generate_schedule()
            for t1, t2 in schedule:
                league.simulate_game(t1, t2)
                
            # 5. Predict & Assign Broadcasts Greedily
            total_season_viewership = broadcaster.assign_schedule(schedule)
            total_viewership_tracker.append(total_season_viewership)
            
            # 6. Print the broadcasted games for the FINAL season
            if season == SEASONS:
                print(f"\n--- Broadcasted Games in Final Season ({policy_name}) ---")
                final_games = broadcaster.final_season_broadcasts
                final_games.sort(key=lambda x: x['score'], reverse=True)
                
                print(f"Total Games Aired: {len(final_games)}")
                for g in final_games[:50]: # Printing Top 50 for brevity
                    team1 = league.get_team(g['t1'])
                    team2 = league.get_team(g['t2'])
                    print(f"Wk {g['w']:>2}, Day {g['d']}, {g['t']:<9} ({g['b']:<9}) | "
                          f"{team1.id} ({team1.power:.1f}) vs {team2.id} ({team2.power:.1f}) | Score: {g['score']:.2f}")

            # 7. End of Season Updates
            policy.calculate_promotion_metrics()
            
            for t in league.teams:
                t.reset_season_stats()
                
        # Mean of total season actual viewership across 50 seasons
        mean_total_viewership = np.mean(total_viewership_tracker)
        results[policy_name] = mean_total_viewership
        viewership_distributions[policy_name] = total_viewership_tracker
        print(f"\nCompleted {SEASONS} seasons. Mean Total Season Viewership: {mean_total_viewership:,.2f}")

    print("\n========================================================")
    print("--- FINAL RESULTS (Mean Total Season Viewership over 50 Seasons) ---")
    for policy, view in results.items():
        print(f"{policy:<30}: {view:,.2f}")

def main():
    coeffs, limits = load_data_from_files()
    
    # 1. Generate the fixed 50-year power timeline
    print("Pre-generating fixed 50-year power trajectories to ensure simulation parity...")
    power_trajectories, historical_baselines = pregenerate_power_trajectories()
    
    policies = {
        "Status Quo": Policy1_StatusQuo,
        "Status Quo + OOC Performance": Policy2_StatusQuo_PerformanceOOC,
        "Conference Tiers": Policy3_ConferenceTiers,
        "Promotion Tiers": Policy4_PromotionTiers
    }
    
    results = {}
    viewership_distributions = {}
    
    # Tracking for the 8-line power graph
    p1_history = {1: [], 2: [], 3: [], 4: []}
    p4_history = {1: [], 2: [], 3: [], 4: []}
    
    for policy_name, PolicyClass in policies.items():
        print(f"\n========================================================")
        print(f"--- Running Simulation for: {policy_name} ---")
        league = League()
        broadcaster = GreedyBroadcaster(coeffs, limits, league)
        total_viewership_tracker = []
        
        # Synchronize Historical Baselines so Tier Bonus calculations align
        for t in league.teams:
            t.historical_power = historical_baselines[t.id]
        
        for season in range(1, SEASONS + 1):
            
            # 1. Apply the exact pre-generated power for this season
            for t in league.teams:
                t.power = power_trajectories[season][t.id]
                t.history_power.append(t.power) 
            
            # 2. Instantiate Policy and Assign Tiers
            policy = PolicyClass(league)
            policy.assign_tiers()
            
            # --- GRAPH 1 TRACKING LOGIC (8 Lines) ---
            if policy_name == "Status Quo":
                conf_powers = defaultdict(list)
                for t in league.teams:
                    conf_powers[t.conf_tag].append(t.power)
                # Sort conferences by average power to consistently track the top 4
                avg_conf_powers = sorted([np.mean(powers) for powers in conf_powers.values()], reverse=True)
                for i in range(4):
                    p1_history[i+1].append(avg_conf_powers[i])
                    
            elif policy_name == "Promotion Tiers":
                tier_powers = defaultdict(list)
                for t in league.teams:
                    tier_powers[t.tier].append(t.power)
                # Tiers 1 through 4 are explicitly the top 4
                for i in range(1, 5):
                    if tier_powers[i]:
                        p4_history[i].append(np.mean(tier_powers[i]))
                    else:
                        p4_history[i].append(0)
            # ----------------------------------------

            # 3. Print Team Powers before simulating the FINAL season
            if season == SEASONS:
                print(f"\n--- Team Powers Going Into Final Season ({policy_name}) ---")
                sorted_teams = sorted(league.teams, key=lambda x: x.power, reverse=True)
                for t in sorted_teams:
                    print(f"{t.id:<18} ({t.conf_tag:<18}) | Power: {t.power:>5.2f} | Tier: {t.tier:>2} | Metric: {t.ranking_metric:>6.2f}")

            # 4. Generate Schedule & Simulate Games
            schedule = policy.generate_schedule()
            for t1, t2 in schedule:
                league.simulate_game(t1, t2)
                
            # 5. Predict & Assign Broadcasts Greedily
            total_season_viewership = broadcaster.assign_schedule(schedule)
            total_viewership_tracker.append(total_season_viewership)
            
            # 6. Print the broadcasted games for the FINAL season
            if season == SEASONS:
                print(f"\n--- Broadcasted Games in Final Season ({policy_name}) ---")
                final_games = broadcaster.final_season_broadcasts
                final_games.sort(key=lambda x: x['score'], reverse=True)
                
                print(f"Total Games Aired: {len(final_games)}")
                for g in final_games[:50]: 
                    team1 = league.get_team(g['t1'])
                    team2 = league.get_team(g['t2'])
                    print(f"Wk {g['w']:>2}, Day {g['d']}, {g['t']:<9} ({g['b']:<9}) | "
                          f"{team1.id} ({team1.power:.1f}) vs {team2.id} ({team2.power:.1f}) | Score: {g['score']:.2f}")

            # 7. End of Season Updates
            policy.calculate_promotion_metrics()
            
            for t in league.teams:
                t.reset_season_stats()
                
        # Mean of total season actual viewership across 50 seasons
        mean_total_viewership = np.mean(total_viewership_tracker)
        results[policy_name] = mean_total_viewership
        viewership_distributions[policy_name] = total_viewership_tracker
        print(f"\nCompleted {SEASONS} seasons. Mean Total Season Viewership: {mean_total_viewership:,.2f}")

    print("\n========================================================")
    print("--- FINAL RESULTS (Mean Total Season Viewership over 50 Seasons) ---")
    for policy, view in results.items():
        print(f"{policy:<30}: {view:,.2f}")

    # ==========================================
    # 8. VISUALIZATIONS
    # ==========================================
    
    # 1. Normalize Viewership (Status Quo = 100 for each season)
    sq_views = viewership_distributions["Status Quo"]
    norm_viewership_distributions = {}
    for policy, views in viewership_distributions.items():
        norm_viewership_distributions[policy] = [(v / sq) * 100 for v, sq in zip(views, sq_views)]
    
    # Graph 1: 8-Line Power Comparison (Absolute Power)
    plt.figure(figsize=(12, 7))
    
    # Color coding by rank to easily compare Tier 1 vs Conf 1, etc.
    colors = {1: 'crimson', 2: 'royalblue', 3: 'forestgreen', 4: 'darkorange'}
    
    for i in range(1, 5):
        # Plot Status Quo Conferences (Dashed)
        plt.plot(range(1, SEASONS + 1), p1_history[i], 
                 label=f'Status Quo - h best conference', 
                 color=colors[i], linestyle='--', alpha=0.7, linewidth=2)
        # Plot Promotion Tiers (Solid)
        plt.plot(range(1, SEASONS + 1), p4_history[i], 
                 label=f'Promotion Tier - {i}', 
                 color=colors[i], linestyle='-', alpha=0.9, linewidth=2)

    plt.title('Average Power Rating Comparison: \n(Top 4 Conferences vs Top 4 Hypothetical Promotion Tiers)')
    plt.xlabel('Season')
    plt.ylabel('Average Power Rating')
    
    # Move legend outside the plot so it doesn't cover the lines
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.savefig('power_comparison_8lines.png')
    plt.show()

    # Graph 2: Box and Whisker of Viewership Outcomes (Normalized)
    plt.figure(figsize=(10, 6))
    plot_data = [norm_viewership_distributions[policy] for policy in policies.keys()]
    
    box = plt.boxplot(plot_data, patch_artist=True, labels=policies.keys())
    
    colors_box = ['lightblue', 'lightgreen', 'lightcoral', 'plum']
    for patch, color in zip(box['boxes'], colors_box):
        patch.set_facecolor(color)
        
    plt.title('Relative Viewership Outcomes Across Policies\n(Normalized per season: Status Quo = 100)')
    plt.ylabel('Relative Viewership Index \n (Status Quo = 100)')
    plt.xlabel('Scheduling Policy')
    plt.xticks(rotation=15)
    
    # Reference line at 100
    plt.axhline(y=100, color='gray', linestyle='--', linewidth=1.5, alpha=0.8) 
    
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.savefig('viewership_boxplot_normalized.png')
    plt.show()

if __name__ == "__main__":
    main()
```
